```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("global_conflicts_dataset.csv")
```

```python
df
```

Horizontal Bar Chart


```python
plt.figure(figsize=(10,5))

plt.hist(
    df["Duration_Days"].dropna(),
    bins=30,
    rwidth=0.8
)

plt.title("Distribution of Conflict Duration")
plt.xlabel("Duration Days")
plt.ylabel("Frequency")

plt.show()
```

Scatter Plot


```python
plt.Figure(figsize=(10,6))

plt.scatter(
    df["Duration_Days"],
    df["Economic_Loss_USD_Billions"],
    alpha=0.5
)
plt.title("Conflict Duration Vs Economic Loss")
plt.xlabel("Duration Days")
plt.ylabel("Economic Loss (USD Billions)")
plt.grid()

plt.show()
```

Horizontal Bar Chart


```python
Top_counties = df["Country_A"].value_counts().head(10)

plt.Figure(figsize=(10,6))
plt.barh(
    Top_counties.index,
    Top_counties.values
)

plt.title("Top 10 countries by conflicts Occurence")
plt.xlabel("Number of Conflicts")
plt.ylabel("Country")

plt.gca().invert_yaxis()

plt.show()
```
